```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal as sig
import scipy.io as sio
from scipy.io import wavfile

# --- Estética de los gráficos ---------------------------------------------
plt.rcParams.update({
    "figure.figsize": (13, 4),
    "axes.grid":      True,
    "grid.alpha":     0.25,
    "axes.titlesize": 12,
    "axes.titleweight": "bold",
    "font.size":      11,
    "lines.linewidth": 1.3,
})

# Paleta consistente por señal
COL = {"ECG": "#c0392b", "PPG": "#2980b9", "Silbido": "#27ae60",
       "Cucaracha": "#8e44ad", "Bonus": "#e67e22", "Seno": "#16a085"}
```

## Utilidades de TS anteriores

Generador de señales (de la TS1) — lo usamos para validar el método y para el bonus.

```python
class GeneradorSenales:
    """Generador de señales periódicas reutilizado de las TS anteriores."""

    @staticmethod
    def _resolver_parametros(ff, n, fs):
        if fs is None and n is None:
            fs = ff * 2
            n = int(fs)
        elif fs is None:
            fs = n
        elif n is None:
            n = int(fs)
        return int(n), float(fs)

    @staticmethod
    def _generar_tiempo(n, fs):
        return (np.arange(n) / fs).reshape(-1, 1)

    @staticmethod
    def generar_seno(vmax=1, dc=0, ff=1, ph=0, n=None, fs=None):
        n, fs = GeneradorSenales._resolver_parametros(ff, n, fs)
        tt = GeneradorSenales._generar_tiempo(n, fs)
        xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)
        return tt, xx

    @staticmethod
    def generar_cuadrada(vmax=1, dc=0, ff=1, ph=0, n=None, fs=None):
        n, fs = GeneradorSenales._resolver_parametros(ff, n, fs)
        tt = GeneradorSenales._generar_tiempo(n, fs)
        xx = dc + vmax * np.sign(np.sin(2 * np.pi * ff * tt + ph))
        return tt, xx
```

## Marco teórico y herramientas

**Método de Welch.** Se parte la señal en $K$ segmentos (con solapamiento), se ventanea cada
uno, se calcula su periodograma y se **promedian**. Promediar reduce la varianza del estimador
a costa de resolución frecuencial: a mayor $K$, segmentos más cortos → espectro más *suave* pero
menos detallado. Es el clásico compromiso **sesgo ↔ varianza**.

**Criterio de ancho de banda.** Definimos el BW como la banda $[f_{inf}, f_{sup}]$ que concentra
un porcentaje dado de la **potencia total** (acá $98\%$). Trabajamos sobre la potencia acumulada
normalizada
$$P_{acum}(f)=\frac{\sum_{f'\le f} \hat{S}(f')}{\sum \hat{S}(f')}$$
y tomamos los cruces por $\tfrac{1-p}{2}$ y $1-\tfrac{1-p}{2}$. Quitamos siempre la componente de
continua (`detrend='constant'`) para que la media no contamine la estimación.

```python
def estimar_psd_welch(x, fs, K=8, detrend="constant"):
    """PSD por Welch. Divide la señal en ~K segmentos (Hann, 50% solapamiento)."""
    x = np.asarray(x).ravel().astype(float)
    nperseg = max(256, len(x) // K)
    f, Pxx = sig.welch(x, fs=fs, window="hann", nperseg=nperseg, detrend=detrend)
    return f, Pxx


def ancho_de_banda(f, Pxx, pot=0.98):
    """Banda [f_inf, f_sup] que concentra `pot` de la potencia total."""
    p_acum = np.cumsum(Pxx)
    p_acum /= p_acum[-1]
    lo = (1 - pot) / 2
    f_inf = f[np.searchsorted(p_acum, lo)]
    f_sup = f[np.searchsorted(p_acum, 1 - lo)]
    return f_inf, f_sup


resultados = []   # acumulador para la tabla comparativa


def analizar_senal(x, fs, nombre, unidad="u.a.", color="C0",
                   K=8, pot=0.98, f_max=None, t_zoom=None):
    """Grafica señal en tiempo + PSD (Welch) con el BW sombreado y registra el resultado."""
    x = np.asarray(x).ravel().astype(float)
    N = len(x)
    t = np.arange(N) / fs

    f, Pxx = estimar_psd_welch(x, fs, K=K)
    Pxx_db = 10 * np.log10(Pxx + 1e-30)
    f_inf, f_sup = ancho_de_banda(f, Pxx, pot)
    bw = f_sup - f_inf
    f_pico = f[np.argmax(Pxx)]

    fig, (ax_t, ax_f) = plt.subplots(1, 2, figsize=(13, 4))

    # --- dominio temporal ---
    if t_zoom is not None:
        m = t <= t_zoom
        ax_t.plot(t[m], x[m], color=color, lw=0.9)
    else:
        ax_t.plot(t, x, color=color, lw=0.9)
    ax_t.set(title=f"{nombre} — dominio temporal",
             xlabel="Tiempo [s]", ylabel=f"Amplitud [{unidad}]")

    # --- dominio frecuencial ---
    ax_f.plot(f, Pxx_db, color=color)
    ax_f.axvspan(f_inf, f_sup, color=color, alpha=0.12,
                 label=f"BW ({pot:.0%}) = {bw:,.1f} Hz")
    for fc in (f_inf, f_sup):
        ax_f.axvline(fc, color="crimson", ls="--", lw=1)
    if f_max:
        ax_f.set_xlim(0, f_max)
    ax_f.set(title=f"{nombre} — PSD (Welch, K={K})",
             xlabel="Frecuencia [Hz]", ylabel="PSD [dB/Hz]")
    ax_f.legend(loc="upper right")
    fig.tight_layout()
    plt.show()

    print(f"  fs = {fs:,} Hz   |   N = {N:,} muestras   |   duración = {N/fs:.2f} s")
    print(f"  Pico espectral en {f_pico:,.2f} Hz")
    print(f"  Banda ocupada ({pot:.0%} de la potencia): "
          f"{f_inf:,.2f} – {f_sup:,.2f} Hz   →   BW ≈ {bw:,.1f} Hz")

    resultados.append(dict(Señal=nombre, **{"fs [Hz]": fs}, N=N,
                           **{"Dur. [s]": round(N/fs, 2),
                              "f_pico [Hz]": round(f_pico, 2),
                              "f_inf [Hz]": round(f_inf, 2),
                              "f_sup [Hz]": round(f_sup, 2),
                              "BW [Hz]": round(bw, 1)}))
    return f, Pxx
```

# 1) Estimación de la PSD de cada señal

## 🫀 ECG — Electrocardiograma

Registro tomado durante una **prueba de esfuerzo**, muestreado a $f_s = 1000\,$Hz.
La energía del ECG se concentra en frecuencias bajas: la onda de pulso ronda ~1 Hz y los
complejos QRS aportan contenido hasta unas pocas decenas de Hz.

```python
fs_ecg = 1000  # Hz
ecg = np.load("ecg_sin_ruido.npy")

f_ecg, P_ecg = analizar_senal(ecg, fs_ecg, "ECG", unidad="adc",
                              color=COL["ECG"], f_max=60, t_zoom=10)
```

## 🩸 PPG — Pletismografía

Señal óptica de pulso registrada en **reposo**, muestreada a $f_s = 400\,$Hz.
Al estar en reposo, el contenido espectral es aún más bajo que el del ECG (la frecuencia
cardíaca y sus primeros armónicos).

```python
fs_ppg = 400  # Hz
ppg = np.load("ppg_sin_ruido.npy")

f_ppg, P_ppg = analizar_senal(ppg, fs_ppg, "PPG", unidad="u.a.",
                              color=COL["PPG"], f_max=20, t_zoom=10)
```

## 🎵 Audio — Silbido

Melodía silbada, grabada a $f_s = 48\,$kHz. Un silbido es casi una senoidal pura que se
desplaza en frecuencia siguiendo la melodía, así que esperamos un contenido concentrado en
una banda relativamente angosta en altas frecuencias (varios kHz).

```python
fs_sil, silbido = wavfile.read("silbido.wav")

f_sil, P_sil = analizar_senal(silbido, fs_sil, "Silbido", unidad="u.a.",
                              color=COL["Silbido"], f_max=12000, t_zoom=0.5)
```

## 🎺 Audio — *La cucaracha*

La misma melodía pero con timbre más rico (más armónicos), grabada a $f_s = 48\,$kHz.
Al tener más armónicos esperamos una banda más extendida que la del silbido puro.

```python
fs_cuc, cucaracha = wavfile.read("la cucaracha.wav")

f_cuc, P_cuc = analizar_senal(cucaracha, fs_cuc, "Cucaracha", unidad="u.a.",
                              color=COL["Cucaracha"], f_max=8000, t_zoom=0.5)
```

## 🔬 Validación del método: el compromiso de Welch

Antes de cerrar, verificamos el método sobre una **senoidal pura de 50 Hz** con potencia
unitaria. Barriendo $K$ vemos el compromiso clásico: con $K$ chico el pico es angosto y
nítido (buena resolución, mucha varianza); con $K$ grande el espectro se suaviza (poca
varianza) pero el pico se ensancha (peor resolución).

```python
N, fs = 45000, 45000
A = np.sqrt(2)  # potencia unitaria
_, xx = GeneradorSenales.generar_seno(vmax=A, ff=50, n=N, fs=fs)
xx = xx.ravel()

fig, ax = plt.subplots(figsize=(11, 5))
for K in [1, 5, 20, 100]:
    f, P = estimar_psd_welch(xx, fs, K=K)
    ax.plot(f, 10 * np.log10(P + 1e-30), label=f"K = {K}", alpha=0.85)
ax.set(title="Welch sobre seno de 50 Hz — efecto de K",
       xlabel="Frecuencia [Hz]", ylabel="PSD [dB/Hz]", xlim=(0, 120))
ax.legend()
plt.tight_layout()
plt.show()
```

# 2) Tabla comparativa de anchos de banda

Reunimos en una sola tabla las estimaciones de todas las señales analizadas.

```python
tabla = pd.DataFrame(resultados).set_index("Señal")
tabla.style.format("{:,.2f}").set_caption(
    "Resumen espectral — BW estimado al 98% de la potencia")
```

**Lectura de la tabla.** Las señales fisiológicas (ECG y PPG) son de **banda baja**:
toda su potencia útil cabe por debajo de ~30 Hz, y la PPG en reposo es aún más angosta (~5 Hz).
Las señales de audio viven varios órdenes de magnitud más arriba: el **silbido** ocupa una banda
de varios kHz centrada en altas frecuencias (tono agudo), mientras que *la cucaracha*, con timbre
más rico en armónicos, reparte su energía en una banda más baja pero igualmente amplia. Esto
justifica por qué cada tipo de señal se digitaliza con frecuencias de muestreo tan distintas
(cientos de Hz para biopotenciales, decenas de kHz para audio).

# Bonus 💎 — Una señal nueva: la red eléctrica (zumbido de 50 Hz)

**Origen.** Proponemos un tipo de señal no analizado antes: la **tensión de línea domiciliaria**
(*mains hum*). En Argentina la red entrega una senoidal de **50 Hz**, pero por las no
linealidades de transformadores y cargas aparecen **armónicos** (150 Hz, 250 Hz, …) y algo de
ruido de fondo. Es el típico interferente que se *cuela* en mediciones de ECG/PPG.

**Digitalización.** En un caso real se captura con una sonda atenuadora + conversor A/D
(p. ej. una placa de adquisición o la entrada de audio de la PC). Acá la **sintetizamos
digitalmente** muestreando a $f_s = 2000\,$Hz un modelo de 50 Hz + armónicos impares + ruido
blanco, lo que reproduce fielmente el espectro de línea: un **espectro de líneas** (muy
angosto) en contraste con las señales de banda continua anteriores.

```python
fs_red = 2000          # Hz
dur = 5                  # s
t = np.arange(fs_red * dur) / fs_red
rng = np.random.default_rng(42)

# 50 Hz + armónicos impares (amplitud decreciente) + ruido de fondo
red = (1.0 * np.sin(2*np.pi*50*t)
       + 0.30 * np.sin(2*np.pi*150*t)
       + 0.15 * np.sin(2*np.pi*250*t)
       + 0.07 * np.sin(2*np.pi*350*t)
       + 0.05 * rng.standard_normal(t.size))

f_red, P_red = analizar_senal(red, fs_red, "Bonus", unidad="V",
                              color=COL["Bonus"], f_max=500, t_zoom=0.1)
```

Como se ve, a diferencia de las señales fisiológicas y de audio (espectros *continuos*),
la red eléctrica es un **espectro de líneas**: casi toda la potencia está en 50 Hz con picos
secundarios en los armónicos impares. El criterio del 98% captura la fundamental y los primeros
armónicos. *(Más adelante podemos reemplazar esta señal sintética por una grabación real de la
entrada de audio si querés.)*

## Apéndice — Comparación con Blackman‑Tukey

El método de **Blackman‑Tukey** estima la PSD como la transformada de Fourier de la
autocorrelación **ventaneada**. Acá lo aplicamos al ECG y lo superponemos con Welch para ver
que ambos métodos coinciden en la forma general del espectro (Welch sale más suave por el
promediado; BT, más detallado pero con más varianza).

```python
def blackman_tukey(x, M=None):
    """PSD por Blackman-Tukey: FFT de la autocorrelacion ventaneada (Hayes)."""
    x = np.asarray(x).ravel().astype(float)
    x = x - x.mean()
    N = len(x)
    if M is None:
        M = N // 5
    r_len = 2 * M - 1
    xx = x[:r_len]
    r = np.correlate(xx, xx, mode="same") / r_len
    Px = np.abs(np.fft.fft(r * sig.windows.blackman(r_len), n=N))
    return Px[:N // 2]


ecg0 = np.load("ecg_sin_ruido.npy").astype(float)
P_bt = blackman_tukey(ecg0, M=len(ecg0) // 15)
f_bt = np.linspace(0, fs_ecg / 2, len(P_bt))

# Welch para comparar (mismo eje)
fw, Pw = estimar_psd_welch(ecg0, fs_ecg, K=8)

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(f_bt, 10*np.log10(P_bt / P_bt.max() + 1e-12), color="#7f8c8d",
        alpha=0.8, label="Blackman-Tukey")
ax.plot(fw, 10*np.log10(Pw / Pw.max() + 1e-12), color=COL["ECG"],
        lw=1.6, label="Welch")
ax.set(title="ECG — Blackman-Tukey vs Welch (PSD normalizada)",
       xlabel="Frecuencia [Hz]", ylabel="PSD [dB] (norm.)", xlim=(0, 60))
ax.legend()
plt.tight_layout()
plt.show()
```

---
### Conclusiones

- **Welch** resultó práctico y robusto para estimar la PSD de señales reales muy distintas
  entre sí, controlando la varianza con el parámetro $K$.
- El **ancho de banda al 98%** ordena claramente las señales: PPG ⟶ ECG ⟶ audio,
  cubriendo desde unos pocos Hz hasta varios kHz.
- El **bonus** mostró el caso opuesto a las señales biológicas: un espectro de líneas
  (red eléctrica) en lugar de un espectro continuo.
- **Blackman‑Tukey** confirmó los resultados de Welch, validando la estimación.
